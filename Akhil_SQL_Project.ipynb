```python
query7 = '''
SELECT
    a.Customer_id,
    a.Account_Number,
    a.Account_type,
    a.Balance_amount +
    COALESCE(
        SUM(
            CASE
                WHEN t.Transaction_amount < 0
                THEN t.Transaction_amount
                ELSE 0
            END
        ), 0
    ) AS deducted_balance_amount
FROM Bank_Account_Details a
LEFT JOIN BANK_ACCOUNT_TRANSACTION t
    ON a.Account_Number = t.Account_Number
WHERE a.Relationship_type = 'P'
GROUP BY
    a.Customer_id,
    a.Account_Number,
    a.Account_type,
    a.Balance_amount;
'''
```

```python
df7 = pd.read_sql_query(query7, conn)
display(df7)
```

```python
query10 = '''
SELECT DISTINCT
    s.Customer_id,
    s.Account_Number AS Savings_Account,
    c.Account_Number AS Credit_Card,
    a.Account_Number AS Add_on_Credit_Card
FROM Bank_Account_Relationship_Details s
JOIN Bank_Account_Relationship_Details c
    ON s.Account_Number = c.Linking_Account_Number
JOIN Bank_Account_Relationship_Details a
    ON c.Account_Number = a.Linking_Account_Number
WHERE s.Account_type = 'SAVINGS'
AND c.Account_type = 'Credit Card'
AND a.Account_type = 'Add-on Credit Card';
'''
```

```python
query11 = '''
SELECT
    s.Account_Number AS Savings_Account,
    c.Account_Number AS Credit_Card,
    SUM(t.Transaction_amount) AS Credit_Aggregate
FROM Bank_Account_Relationship_Details s
JOIN Bank_Account_Relationship_Details c
    ON s.Account_Number = c.Linking_Account_Number
JOIN BANK_ACCOUNT_TRANSACTION t
    ON c.Account_Number = t.Account_Number
WHERE s.Account_type = 'SAVINGS'
AND c.Account_type = 'Credit Card'
GROUP BY s.Account_Number, c.Account_Number;
'''
```
